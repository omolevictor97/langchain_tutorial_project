```python
def some_fn(a, b, *args):
    return a, b

some_fn(4,5, 6, 7)
```

```python
def log_alert(level, message, *details):
    """
    Logs system alerts with a flexible number of additional context details.
    """
    print(f"[{level.upper()}] {message}")
    # for if statement to return True, you are either pass in a non-empty string, an int but not 0
    # you must pass an iterable (list, tuple, set, dict, string), they must be non-empty
    # *args in any functional is tuple
    if details:
        print("Additional Context:")
        for idx, detail in enumerate(details, start=1):
            print(f'Detail {idx}: {detail}')

log_alert("Critical", "Payment gateway timeout", "User_ID : 1044", "Order_ID:TXN-882", "IP_Address", "Error_Code:504")
```
